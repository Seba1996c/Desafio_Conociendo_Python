1. Calculando el salario semanal

Un vendedor gana un salario base de $500 más una comisión del 10% sobre sus
ventas totales. Escribe un programa donde se pueda ingresar el monto de las ventas
totales del vendedor y calcule su salario total.

```python
# Creamos tres variables, que guarden los datos solicitados

ventas_totales = float(input("Cuantos fueron las ventas totales : "))
salario_base = 500
comision = (ventas_totales * 10) / 100

#Creamos una variable que sume las tres variables antes mencionadas
salario_total = salario_base + comision

print(f"El salario total es : ${salario_total}")
```

2. Bienvenido a la frutería virtual

En este ejercicio, vamos a crear un diccionario vacío llamado catalogo_frutas {'fruta':
precio} y agregar algunas frutas con sus precios. Luego, le pediremos al usuario que
ingrese el nombre de una fruta y verificaremos si esa fruta está en el diccionario. Si
la fruta está en el diccionario, mostraremos el precio al usuario. Si la fruta no está en
el diccionario, le diremos al usuario que no se encuentra esa fruta en nuestra tienda.

```python
catalogo_frutas = {}
```

```python
catalogo_frutas
```

```python
type(catalogo_frutas)
```

```python
# Creamos nuestro diccionario con frutas y sus precios

catalogo_frutas = {'manzana':500,'banana':200,'pera':350,'naranja':600}
```

```python
# Validamos que se creo un dict

catalogo_frutas
type(catalogo_frutas)
```

```python
# Visulalizamos como queda nuestro dict

catalogo_frutas
```

```python
fruta = input("Ingresa el nombre de la fruta que deseas buscar: ").lower()
```

```python
# Buscamos la fruta en el catalogo frutas, donde revise si esta o no

if fruta in catalogo_frutas:
    print(f"El precio de la {fruta} es {catalogo_frutas[fruta]} pesos.")
else:
    print("Lo sentimos, esa fruta no se encuentra en nuestra tienda.")
```

3. Bienvenido al casino
   
En este juego, tienes que lanzar dos dados y el programa mostrará los valores
obtenidos en cada dado y la suma de ambos valores. Debes generar dos números
aleatorios entre 1 y 6 que representan los valores obtenidos en cada dado usando la
librería random. Luego, el programa debe mostrar los valores obtenidos en cada
dado y la suma de ambos valores en la pantalla.

```python
# Importamos la libreria random para poder generar numeros aleatoreos
import random
```

```python
# Definimos nuestros dos datos, del 1 al 6
dado_1 = random.randint(1,6)
dado_2 = random.randint(1,6)

# Sumamos los dos datos e imprimimos 
suma_dado = dado_1 + dado_2
```

```python
# Imprmir los valores solicitados

print(f"El valor de mi primer dado es : {dado_1}")
print(f"El valor de mi segundo dado es : {dado_2}")
print(f"La suma de ambos datos es : {suma_dado}")
```

4 Calculadora de chocolate

Crea un programa que calcule la cantidad de chocolate que puedes comer sin
pasarte de tu límite diario de calorías. El programa debe preguntar al usuario su
límite diario de calorías y la cantidad de calorías que tiene una porción de chocolate.
Luego, debe calcular la cantidad máxima de chocolate que se puede comer sin
exceder el límite diario de calorías y mostrarla en pantalla.

```python
# Definimos dos variables, como int, debido que son numeros, los cuales son la torta de chocolate y las calorias diarias

porcion_torta_chocolate = 345
calorias_diarias = int(input("Cuantas son tus calorias diarias?:"))

# Luego hacemos print(f)
# para que nos indique la cantidad de calorias que tenemos 
# y lo que hace una torta en calorias

print(f"Tu cantidad de calorias diarias es : {calorias_diarias} calorias")
print(f"Una torta de chocolate tiene {porcion_torta_chocolate} calorias")
```

```python
# Creamos la variable max la cantidad máxima de porciones de chocolate que se pueden comer
max_porciones = calorias_diarias // porcion_torta_chocolate

# Mostrar el resultado en pantalla
print(f"Puedes comer hasta {max_porciones} porciones de chocolate sin pasarte de tu límite diario de calorías.")
```
